```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df= pd.read_csv("student exam_score.csv")
df
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
print(df.head(5))
```

```python
df.isnull().sum()
```

```python
df['EthnicGroup'] =df['EthnicGroup'].fillna('not specified')
print(df['EthnicGroup'].value_counts())
```

```python
df.head(5)
```

```python
df['TestPrep'] =df['TestPrep'].fillna('not specified')
print(df['TestPrep'].value_counts())
```

```python
df['ParentMaritalStatus'] =df['ParentMaritalStatus'].fillna('un known')
print(df['ParentMaritalStatus'].value_counts())
```

```python
df['ParentEduc'] =df['ParentEduc'].fillna('not specified')
print(df['ParentEduc'].value_counts())
```

```python
df['WklyStudyHours'] =df['WklyStudyHours'].fillna(0)
print(df['WklyStudyHours'].value_counts())
```

```python
df['IsFirstChild'] =df['IsFirstChild'].fillna('not applicable')
df['PracticeSport'] =df['PracticeSport'].fillna('NA')
df['NrSiblings'] =df['NrSiblings'].fillna('sibilings not specified')
df['TransportMeans'] =df['TransportMeans'].fillna('Transport not specified')
print(df['IsFirstChild'].value_counts())
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
plt.figure(figsize =(3,5))
count = sns.countplot(data = df, x="Gender")
count.bar_label(count.containers[0])
plt.show()
```

```python
ab = df.groupby("ParentEduc").agg({"MathScore":"mean","ReadingScore":"mean","WritingScore":"mean"})
print(ab)
```

```python
df[['MathScore','ReadingScore','WritingScore']].hist(figsize=(10,5))
plt.show()
```

```python

df.boxplot(column='MathScore', by='Gender', figsize=(6,5))
plt.title("Gender vs Mathscore")
plt.show()
```
